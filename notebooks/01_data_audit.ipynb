The response variable is Loan Status, which represents the loan approval outcome.

# Missing-Value Assessment

Missing-value analysis was performed using both the number and percentage of missing observations.

The analysis showed that all 14 variables contain zero missing values. This means that missing-value imputation is not required for the current dataset. However, this finding does not imply that the dataset is free from data-quality problems. Implausible, extreme, or inconsistent values can exist even when all observations are complete.

Consequently, the subsequent data-quality assessment focuses on the validity and plausibility of the observed values.

# Duplicate-Record Assessment

The dataset was also examined for duplicate observations. Therefore, no completely duplicated records were identified.

This indicates that there are no exact duplicate rows requiring removal at this stage. Nevertheless, duplicate applicants cannot be completely ruled out because two applications could contain different values while belonging to the same individual. Such an assessment would require an applicant identifier, which is not included among the 14 variables.

# Descriptive Statistics

Descriptive statistics were examined for the numerical variables.

1 Age
The observed age has:
Mean =27.76 and SD=6.05.
The first quartile is 24 years, the median is 26 years, and the third quartile is 30 years.

However, the minimum and maximum values require particular attention. The minimum age is 20 years, while the maximum is 144 years. An age of 144 years is implausible for a loan applicant and strongly indicates the presence of an extreme or erroneous observation.

This is an important data-quality issue that should be investigated before modelling. The mean age of 27.76 years may also be affected by such extreme observations.

2. Person Income

Applicant income has a mean of approximately is 80,319.05 with a standard deviation of 80,422.50.
The median income is approximately 67,048, while the third quartile is approximately 95,789.

The most important finding is the maximum income is 7,200,766. This is substantially greater than the upper quartile and indicates a highly right-skewed distribution and/or the presence of extreme observations.

The difference between the mean and median also supports the possibility of substantial positive skewness.

The income variable therefore requires further investigation before modelling. Extreme values should not automatically be deleted; instead, their validity and potential influence should be assessed.

3. Employee Experience

Employee experience has a mean of approximately is 5.41 years and a median of 4 years. The minimum value is 0 years, which can be reasonable for applicants with no previous employment experience.

However, the maximum value is
125 years. This is highly implausible and indicates another potential data-quality problem.

Furthermore, employee experience should be considered in relation to applicant age. For example, an applicant aged 20 with 125 years of employment experience would be impossible.

Therefore, a consistency check between Age and Employee Experience should be performed.

4. Loan Amount

The requested loan amount has a mean of approximately is 9,583.16 and a standard deviation of approximately is 6,314.89.

The median loan amount is 8,000, while the first and third quartiles are 5,000 and 12,237.25, respectively.

The observed loan amounts range from  500 to 35,000. Unlike the age, income, and employment-experience variables, the loan amount does not immediately display an obviously impossible maximum. However, its distribution should still be examined graphically because the mean exceeds the median, suggesting some positive skewness.

5. Loan Interest Rate

The loan interest rate has a mean of approximately: 11.01% with a standard deviation of approximately 2.98. The minimum observed rate is 5.42%, while the maximum is 20%.

These values appear plausible within the context of the dataset, although the meaning and generation of this variable require further investigation.

In particular, it must be established whether the interest rate was available at the time the approval decision was made.

If the interest rate was assigned only after a loan was approved, using it as a predictor of approval could introduce data leakage. This issue must therefore be resolved before model development.

6. Loan Percentage

The Loan percentage variable has a mean of approximately: 0.1397 and a standard deviation of approximately: 0.0872. The median is 0.12 and the maximum is 0.66.

This suggests that Loan percentage may be a derived financial feature measuring the requested loan relative to applicant income.

This relationship should be formally verified before feature engineering is undertaken.
